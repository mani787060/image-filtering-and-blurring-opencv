```python
import cv2
import numpy as np
```

```python
# using 'Gaussian_blur' on an image :-

image = cv2.imread("nature image.jpg")

blurred = cv2.GaussianBlur(image, (7,7), 0)

# printing the image 
cv2.imshow("Original Image", image)
cv2.imshow("Blurred Image", blurred)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# using 'Median_blur' on an image :-

image = cv2.imread("rose image.jpg")

blurred = cv2.medianBlur(image, 11)

# printing the image
cv2.imshow("Original image", image)
cv2.imshow("Clean image", blurred)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# sharpening an image :-

image = cv2.imread("rose image.jpg")

sharpened_kernel = np.array([
    [0,-1,0],
    [-1,5,-1],
    [0,-1,0]
])

sharpened = cv2.filter2D(image, -1, sharpened_kernel)

cv2.imshow("Original image", image)
cv2.imshow("Sharpened image", sharpened)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
